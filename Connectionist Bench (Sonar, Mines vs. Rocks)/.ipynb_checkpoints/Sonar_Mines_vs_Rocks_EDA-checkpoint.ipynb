```python
from pathlib import Path
import pandas as pd
import numpy as np

dataset_file = Path(
    r"C:\Users\kavya\Connectionist Bench (Sonar, Mines vs. Rocks)\dataset"
    r"\connectionist+bench+sonar+mines+vs+rocks\sonar.all-data"
)

# Sonar dataset has no header: 60 numeric features + class label
column_names = [f"Feature_{i}" for i in range(1, 61)] + ["Class"]

df = pd.read_csv(dataset_file, header=None, names=column_names)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python
# Dataset structure and data types
print("Dataset Information:")
df.info()

print("\nDataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Statistical summary of numeric features
print("Statistical Summary:")
display(df.describe())

# Class label summary
print("\nClass Distribution:")
print(df["Class"].value_counts())

print("\nClass Percentages:")
display((df["Class"].value_counts(normalize=True) * 100).round(2))
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Total missing values:", missing_values.sum())

print("\nColumns with missing values:")
display(missing_values[missing_values > 0])

# Check duplicate rows
duplicate_count = df.duplicated().sum()
print("\nNumber of duplicate rows:", duplicate_count)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Class")

plt.title("Sonar Dataset - Mines vs. Rocks Class Distribution")
plt.xlabel("Class (M = Mine, R = Rock)")
plt.ylabel("Number of Samples")
plt.show()
```

```python
# Distribution of the first 10 sonar features
features_to_plot = [f"Feature_{i}" for i in range(1, 11)]

df[features_to_plot].hist(
    figsize=(15, 10),
    bins=20,
    edgecolor="black"
)

plt.suptitle("Distribution of First 10 Sonar Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Compare selected sonar features for Mines and Rocks
features_to_compare = [
    "Feature_1",
    "Feature_10",
    "Feature_20",
    "Feature_30",
    "Feature_40",
    "Feature_50",
    "Feature_60"
]

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

for i, feature in enumerate(features_to_compare):
    sns.boxplot(data=df, x="Class", y=feature, ax=axes[i])
    axes[i].set_title(f"{feature} by Class")

# Hide unused subplot
axes[-1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# Correlation between the 60 numeric sonar features
numeric_df = df.drop(columns=["Class"])

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    xticklabels=False,
    yticklabels=False
)

plt.title("Correlation Heatmap of Sonar Features")
plt.xlabel("Features")
plt.ylabel("Features")
plt.show()
```

## Results

Exploratory Data Analysis (EDA) was performed on the Connectionist Bench Sonar Mines vs. Rocks dataset. The dataset was loaded and examined using its structure, data types, statistical summaries, and class distribution.

Missing values and duplicate records were checked. Visualizations were created to understand the distribution of sonar features, compare selected feature values between Mines and Rocks, and examine correlations among the numeric features.

## Conclusion

The EDA process provided an understanding of the Sonar dataset and its feature characteristics. The dataset contains sonar measurements used to distinguish between mines and rocks. The analysis and visualizations provide a foundation for data preprocessing, feature preparation, and classification model development.

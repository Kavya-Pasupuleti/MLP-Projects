```python
from pathlib import Path
import pandas as pd
import numpy as np

# Locate the saved preprocessed dataset
matches = list(Path.cwd().rglob("sonar_preprocessed.csv"))

if not matches:
    raise FileNotFoundError(
        "sonar_preprocessed.csv not found. Check that Notebook 2 saved the file."
    )

cleaned_file = matches[0]
df = pd.read_csv(cleaned_file)

print("Loaded file:", cleaned_file)
print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Input features (60 sonar measurements)
feature_columns = [f"Feature_{i}" for i in range(1, 61)]

X = df[feature_columns].copy()

# Target variable (1 = Mine, 0 = Rock)
y = df["Class"].copy()

print("Feature matrix shape:", X.shape)
print("Target vector shape:", y.shape)

print("\nTarget class distribution:")
display(y.value_counts())

print("\nFeature sample:")
display(X.head())
```

```python
from sklearn.model_selection import train_test_split

# Split dataset: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)

print("\nTraining target distribution:")
display(y_train.value_counts())

print("\nTesting target distribution:")
display(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize scaler
scaler = StandardScaler()

# Fit scaler on training data and transform both sets
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)

print("\nTraining feature means (first 5):")
print(np.round(X_train_scaled.mean(axis=0)[:5], 4))

print("\nTraining feature standard deviations (first 5):")
print(np.round(X_train_scaled.std(axis=0)[:5], 4))
```

```python
from pathlib import Path
import joblib

output_dir = Path.cwd() / "Sonar_Mines_vs_Rocks"
output_dir.mkdir(parents=True, exist_ok=True)

# Save scaled features
pd.DataFrame(X_train_scaled, columns=feature_columns).to_csv(
    output_dir / "X_train_scaled.csv", index=False
)
pd.DataFrame(X_test_scaled, columns=feature_columns).to_csv(
    output_dir / "X_test_scaled.csv", index=False
)

# Save target labels
y_train.to_csv(output_dir / "y_train.csv", index=False, header=["Class"])
y_test.to_csv(output_dir / "y_test.csv", index=False, header=["Class"])

# Save scaler for future use
joblib.dump(scaler, output_dir / "sonar_scaler.pkl")

print("Prepared datasets and scaler saved successfully!")
print("Output folder:", output_dir)
```

## Results

The preprocessed Sonar dataset was divided into input features and the target variable. The 60 sonar measurements were used as input features, while the encoded class label (Mine = 1, Rock = 0) was used as the target.

The dataset was split into training and testing sets using an 80:20 ratio with stratified sampling. StandardScaler was applied to normalize the feature values. The scaler was fitted only on the training data and then used to transform the testing data.

The scaled datasets and fitted scaler were saved for model development.

## Conclusion

Feature preparation was completed successfully. The training and testing datasets are ready for classification algorithms to distinguish between mines and rocks using sonar measurements.

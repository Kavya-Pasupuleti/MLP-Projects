```python
from pathlib import Path
import pandas as pd
import numpy as np

dataset_file = Path(
    r"C:\Users\kavya\Connectionist Bench (Sonar, Mines vs. Rocks)\dataset"
    r"\connectionist+bench+sonar+mines+vs+rocks\sonar.all-data"
)

column_names = [f"Feature_{i}" for i in range(1, 61)] + ["Class"]

df = pd.read_csv(dataset_file, header=None, names=column_names)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Dataset Information:")
df.info()

print("\nMissing values per column:")
display(df.isnull().sum()[df.isnull().sum() > 0])

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Check duplicate rows
duplicate_count = df.duplicated().sum()
print("Duplicate rows found:", duplicate_count)

# Remove duplicates if present
df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

```python
# Separate feature columns and target column
feature_columns = [f"Feature_{i}" for i in range(1, 61)]

# Convert feature values to numeric
df[feature_columns] = df[feature_columns].apply(
    pd.to_numeric, errors="coerce"
)

# Check target labels
print("Class labels found:", df["Class"].unique())

# Check for invalid or missing feature values
print("\nMissing values after numeric conversion:", df.isnull().sum().sum())

print("\nClass distribution:")
display(df["Class"].value_counts())

print("\nUpdated dataset shape:", df.shape)
```

```python
# Fill missing numeric feature values using median
df[feature_columns] = df[feature_columns].fillna(
    df[feature_columns].median()
)

# Encode class labels: M = Mine (1), R = Rock (0)
df["Class"] = df["Class"].map({"M": 1, "R": 0})

# Verify preprocessing
print("Missing values remaining:", df.isnull().sum().sum())
print("\nEncoded class distribution:")
display(df["Class"].value_counts())

print("\nFirst 5 rows after preprocessing:")
display(df.head())
```

```python
output_dir = Path.cwd() / "Sonar_Mines_vs_Rocks"
output_dir.mkdir(parents=True, exist_ok=True)

cleaned_file = output_dir / "sonar_preprocessed.csv"

df.to_csv(cleaned_file, index=False)

print("Preprocessed dataset saved successfully!")
print("Saved at:", cleaned_file)
print("Final dataset shape:", df.shape)
```

## Results

The Sonar Mines vs. Rocks dataset was loaded and inspected. Data types, missing values, and duplicate records were checked. Numeric feature values were validated, and any missing feature values were handled using median imputation.

The class labels were encoded numerically: Mine (M) as 1 and Rock (R) as 0. The cleaned dataset was saved as `sonar_preprocessed.csv`.

## Conclusion

The preprocessing of the Sonar dataset was completed successfully. The dataset is now cleaned, has numeric input features, and contains an encoded target variable. It is ready for feature preparation and classification model development.

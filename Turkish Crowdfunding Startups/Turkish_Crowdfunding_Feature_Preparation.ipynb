```python
from pathlib import Path
import pandas as pd
import numpy as np

# Find the file saved in Notebook 2
matches = list(Path.cwd().rglob("turkish_crowdfunding_preprocessed.csv"))

if not matches:
    raise FileNotFoundError(
        "Could not find turkish_crowdfunding_preprocessed.csv. "
        "Make sure Notebook 2 was run and saved."
    )

file_path = matches[0]
X = pd.read_csv(file_path)

print("Loaded file:", file_path)
print("Dataset shape:", X.shape)
X.head()
```

```python
# Check for missing values
missing_values = X.isnull().sum().sum()

# Check for infinite values
infinite_values = np.isinf(X.select_dtypes(include=np.number)).sum().sum()

print("Missing values:", missing_values)
print("Infinite values:", infinite_values)
print("Duplicate rows:", X.duplicated().sum())

if missing_values == 0 and infinite_values == 0:
    print("\nData is clean and ready for clustering.")
else:
    print("\nData needs further cleaning.")
```

```python
# Keep only numeric features for clustering
X_cluster = X.select_dtypes(include=[np.number]).copy()

print("Original processed dataset shape:", X.shape)
print("Clustering dataset shape:", X_cluster.shape)

# Confirm all values are numeric and finite
print("All values finite:", np.isfinite(X_cluster.to_numpy()).all())

X_cluster.head()
```

```python
from pathlib import Path

output_dir = Path("processed_data")
output_dir.mkdir(exist_ok=True)

output_file = output_dir / "turkish_crowdfunding_cluster_ready.csv"
X_cluster.to_csv(output_file, index=False)

print("Clustering-ready dataset saved to:", output_file)
print("Final dataset shape:", X_cluster.shape)
```

## Result

The preprocessed Turkish Crowdfunding Startups dataset was loaded and checked for missing and infinite values. The numeric features were extracted and prepared as a clustering-ready dataset. The final dataset was saved as `turkish_crowdfunding_cluster_ready.csv`.

## Conclusion

Feature preparation was completed successfully. The dataset is now ready for applying unsupervised clustering algorithms to identify groups of crowdfunding projects with similar characteristics.

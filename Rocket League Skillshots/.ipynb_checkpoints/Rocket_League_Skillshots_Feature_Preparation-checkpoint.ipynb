```python
from pathlib import Path
import pandas as pd
import numpy as np

matches = list(Path.cwd().rglob("rocket_league_skillshots_cleaned.csv"))

if not matches:
    raise FileNotFoundError(
        "Could not find rocket_league_skillshots_cleaned.csv. "
        "Make sure Notebook 2 was run and saved."
    )

file_path = matches[0]
df_clean = pd.read_csv(file_path)

print("Loaded file:", file_path)
print("Dataset shape:", df_clean.shape)

df_clean.head()
```

```python
feature_columns = [
    "BallAcceleration",
    "Time",
    "DistanceWall",
    "DistanceCeil",
    "DistanceBall",
    "PlayerSpeed",
    "BallSpeed"
]

action_columns = [
    "up",
    "accelerate",
    "slow",
    "goal",
    "left",
    "boost",
    "camera",
    "down",
    "right",
    "slide",
    "jump"
]

# Keep only columns available in the cleaned dataset
feature_columns = [col for col in feature_columns if col in df_clean.columns]
action_columns = [col for col in action_columns if col in df_clean.columns]

X = df_clean[feature_columns].copy()
y = df_clean[action_columns].copy()

print("Input features shape:", X.shape)
print("Target labels shape:", y.shape)

print("\nInput feature columns:", feature_columns)
print("\nTarget label columns:", action_columns)

display(X.head())
display(y.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)
```

```python
from pathlib import Path

output_dir = Path("processed_data")
output_dir.mkdir(exist_ok=True)

# Save scaled features
pd.DataFrame(
    X_train_scaled, columns=feature_columns
).to_csv(output_dir / "rocket_league_X_train.csv", index=False)

pd.DataFrame(
    X_test_scaled, columns=feature_columns
).to_csv(output_dir / "rocket_league_X_test.csv", index=False)

# Save target labels
y_train.to_csv(output_dir / "rocket_league_y_train.csv", index=False)
y_test.to_csv(output_dir / "rocket_league_y_test.csv", index=False)

print("Prepared training and testing datasets saved successfully.")
```

## Result

The cleaned Rocket League Skillshots dataset was divided into input features and multiple action target labels. The data was split into training and testing sets using an 80:20 ratio. StandardScaler was applied to the gameplay measurement features, and the prepared datasets were saved for classification.

## Conclusion

Feature preparation was completed successfully. The scaled training and testing features, along with their corresponding action labels, are ready for training and evaluating a multi-label classification model.

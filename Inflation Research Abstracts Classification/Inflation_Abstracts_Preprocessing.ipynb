```python
import pandas as pd
import json
import re
```

```python
with open(
    "dataset/inflation+research+abstracts+classification/classified_abstracts.json",
    "r",
    encoding="utf-8"
) as file:
    data = json.load(file)

df = pd.DataFrame(data)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df = df[["Abstract", "Label"]]

print("Selected columns:")
print(df.columns.tolist())

df.head()
```

```python
import re
```

```python
print("Duplicates before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after cleaning:", df.duplicated().sum())
print("Updated shape:", df.shape)
```

```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"<[^>]+>", " ", text)       # Remove HTML tags
    text = re.sub(r"http\S+|www\S+", " ", text) # Remove URLs
    text = re.sub(r"[^a-zA-Z\s]", " ", text)   # Keep letters and spaces
    text = re.sub(r"\s+", " ", text).strip()   # Remove extra spaces
    return text

df["Abstract"] = df["Abstract"].apply(clean_text)

df[["Abstract", "Label"]].head()
```

```python
df[["Abstract", "Label"]].head(10)
```

```python
empty_abstracts = df["Abstract"].str.strip().eq("").sum()

print("Empty abstracts after cleaning:", empty_abstracts)
```

```python
# Remove abstracts that became empty after cleaning
df = df[df["Abstract"].str.strip() != ""]

# Convert labels to integers
df["Label"] = df["Label"].astype(int)

# Check the cleaned dataset
print("Dataset shape after preprocessing:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
print("\nLabel distribution:")
print(df["Label"].value_counts())

# Display sample cleaned abstracts
display(df.head())
```

## Result
The Inflation Research Abstracts Classification dataset was preprocessed successfully. Missing and duplicate records were handled, and the abstract text was cleaned by converting it to lowercase and removing HTML tags, URLs, punctuation, numbers, and extra spaces. Empty abstracts were removed, and the target labels were converted to integers. The cleaned dataset was saved as a CSV file.

## Conclusion
The textual data was cleaned and prepared for machine learning. The preprocessed dataset contains the abstract text and binary classification labels. It is ready for the train-test splitting and 
classification stages.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

print("Libraries imported successfully!")
```

```python
files = list(Path.cwd().rglob("onehr.data"))

if not files:
    print("Dataset file not found. Check the notebook folder.")
else:
    data_path = files[0]
    df = pd.read_csv(data_path, header=None, na_values=["?"])

    print("Dataset loaded successfully!")
    print("Dataset shape:", df.shape)
    display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())

print("\nStatistical Summary:")
display(df.describe())
```

```python
# Check missing values in each column
print("Missing Values:")
display(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
plt.figure(figsize=(14, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Missing Values in Ozone Dataset")
plt.xlabel("Features")
plt.ylabel("Records")
plt.show()
```

```python
numeric_data = df.select_dtypes(include=np.number)

numeric_data.iloc[:, :12].hist(
    figsize=(16, 12),
    bins=30,
    edgecolor="black"
)

plt.suptitle("Distribution of First 12 Numeric Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 6))


sns.boxplot(data=numeric_data.iloc[:, :12])

plt.title("Boxplots of First 12 Numeric Features")
plt.xlabel("Features")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.show()
```

```python
target = df.iloc[:, -1]

print("Class Distribution:")
display(target.value_counts(dropna=False))

plt.figure(figsize=(6, 4))
sns.countplot(x=target)

plt.title("Ozone Class Distribution")
plt.xlabel("Ozone Class")
plt.ylabel("Number of Records")
plt.show()
```


## Result

Exploratory Data Analysis (EDA) was performed on the UCI Ozone Level Detection dataset. The dataset was examined using its shape, data types, statistical summaries, and sample records. Missing values and duplicate records were checked. Histograms, boxplots, a missing-value heatmap, and a class distribution plot were used to visualize the data and identify potential outliers and class imbalance.

## Conclusion

The EDA process provided an understanding of the structure, quality, and distribution of the Ozone Level Detection dataset. The analysis revealed missing values and an imbalance in the target classes. These findings will guide the next stage of data preprocessing before applying machine learning classification algorithms.

```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.impute import SimpleImputer

print("Libraries imported successfully!")
```

```python
files = list(Path.cwd().rglob("onehr.data"))

if not files:
    print("Dataset file not found. Check the notebook folder.")
else:
    data_path = files[0]
    df = pd.read_csv(data_path, header=None, na_values=["?"])

    print("Dataset loaded successfully!")
    print("Shape:", df.shape)
    display(df.head())
```

```python
# Assign column names
df.columns = ["Date"] + [f"Feature_{i}" for i in range(1, 73)] + ["Ozone_Class"]

# Remove rows where the target class is missing
df = df.dropna(subset=["Ozone_Class"])

# Convert target to integer labels
df["Ozone_Class"] = df["Ozone_Class"].astype(int)

# Separate input features and target
X = df.drop(columns=["Date", "Ozone_Class"])
y = df["Ozone_Class"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget class distribution:")
display(y.value_counts())
```

```python
# Create median imputer
imputer = SimpleImputer(strategy="median")

# Impute missing values in the features
X_imputed = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns,
    index=X.index
)

print("Missing values before imputation:", X.isnull().sum().sum())
print("Missing values after imputation:", X_imputed.isnull().sum().sum())

display(X_imputed.head())
```

```python
# Combine imputed features and target
processed_df = X_imputed.copy()
processed_df["Ozone_Class"] = y

# Save the preprocessed dataset
output_path = Path.cwd() / "ozone_preprocessed.csv"
processed_df.to_csv(output_path, index=False)

print("Preprocessed dataset saved successfully!")
print("Saved file:", output_path)
print("Dataset shape:", processed_df.shape)

display(processed_df.head())
```


## Result

The UCI Ozone Level Detection dataset was loaded and prepared for machine learning. The date column was excluded from the input features, and the target column was identified as `Ozone_Class`. Records with missing target values were removed. Missing values in the numerical features were handled using median imputation. The cleaned features and target class were combined and saved as `ozone_preprocessed.csv`.

## Conclusion

Data preprocessing was successfully completed. Missing feature values were imputed, and the dataset was prepared with 72 input features and one target class. The preprocessed dataset was saved for use in the train-test splitting and classification stages.
